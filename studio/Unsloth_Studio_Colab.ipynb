On Colab, `start()` auto-opens a shareable Cloudflare link because the in-cell proxy iframe often stays blank. Open Unsloth with that link above the ready card. To skip the tunnel and try the in-notebook proxy iframe only, use `start(cloudflare=False)`.

A shareable link needs a login so it is not open to everyone. If you do not set a password, Unsloth auto-generates a strong one and shows it once in the cell output below as `username: unsloth` and `password: ...`. Copy that password into the Unsloth login page. The server keeps it out of its own logs and out of the shared link, but a notebook SAVES its cell output: Colab autosaves to Drive, and an exported or shared copy carries that output with it. So treat the password as part of this notebook until you clear the output of the cell below (or change the password with `unsloth studio reset-password`). To choose your own instead, set the `UNSLOTH_STUDIO_PASSWORD` environment variable before calling `start()`.

```python
import sys
import os

# Robustly change to a known stable directory first to avoid FileNotFoundError on os.getcwd()
%cd /content
print(f"Working directory after initial cd to /content: {os.getcwd()}")

# Ensure the unsloth repository is cloned and set up
if os.path.exists("/content/unsloth"):
    print("Removing existing /content/unsloth directory to ensure a clean setup.")
    !rm -rf /content/unsloth

print("Cloning unsloth repository...")
!git clone --depth 1 --branch main https://github.com/unslothai/unsloth.git

# Change directory for setup script
print(f"Current working directory before cd to unsloth: {os.getcwd()}")
%cd /content/unsloth
print(f"Current working directory after cd to unsloth: {os.getcwd()}")

print("Running setup script...")
!chmod +x studio/setup.sh && ./studio/setup.sh --local

# --- Diagnostics for ModuleNotFoundError ---
# Verify existence of the 'colab.py' file right before import
colab_file_path = "/content/unsloth/studio/backend/colab.py"
print(f"Checking for file: {colab_file_path}")
if os.path.exists(colab_file_path):
    print(f"Verified: '{colab_file_path}' exists.")
    print(f"Contents of directory {os.path.dirname(colab_file_path)}: {os.listdir(os.path.dirname(colab_file_path))}")
    # Read the content of colab.py
    print(f"\nContent of {colab_file_path}:")
    with open(colab_file_path, 'r') as f:
        print(f.read())
else:
    print(f"CRITICAL ERROR: '{colab_file_path}' does NOT exist after setup. Module cannot be imported.")

# Clean and add the backend directory to sys.path
# Remove existing entries to avoid duplicates and ensure the correct order
backend_path = "/content/unsloth/studio/backend"
sys.path = [p for p in sys.path if p != backend_path] # Remove all existing occurrences
sys.path.insert(0, backend_path) # Insert it at the beginning

# Print sys.path to verify for debugging purposes
print("Current sys.path:")
for i, p in enumerate(sys.path):
    print(f"  {i}: {p}")

print(f"Current working directory before import: {os.getcwd()}")
print("Attempting to import 'colab' module and check for 'start'...")

# Clear 'colab' from sys.modules to force a fresh import
if 'colab' in sys.modules:
    print("\n'colab' found in sys.modules. Removing to force re-import.")
    del sys.modules['colab']

try:
    import colab
    print("Successfully imported 'colab' module.")
    print(f"Imported 'colab' module from: {colab.__file__}")

    if hasattr(colab, 'start'):
        print("Found 'start' attribute in 'colab' module. Attempting to call colab.start().")
        colab.start()
    else:
        print("Error: 'start' attribute not found in 'colab' module after import.")
        print(f"Available attributes in 'colab' module: {dir(colab)}")

except Exception as e:
    print(f"Error during module import or function call: {e}")
```
